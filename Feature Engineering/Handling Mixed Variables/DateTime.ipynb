```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('google_playstore_2026.csv')
```

```python
df.head()
```

```python
df = df[['App', 'Category', 'Size', 'Last Updated']]
```

```python
df.sample()
```

```python
df['Last Updated'] = pd.to_datetime(df['Last Updated'])
```

```python
df.sample()
```

```python
df['Update Year'] = df['Last Updated'].dt.year
df.head()
```

```python
df['Update Month'] = df['Last Updated'].dt.month
df.head()
```

```python
df['Update Month'] = df['Last Updated'].dt.month_name()
df.head()
```

```python
df['Update Date'] = df['Last Updated'].dt.day
df.head()
```

```python
df['Update Day of week'] = df['Last Updated'].dt.dayofweek
df.head()
```

```python
df['Update Day'] = df['Last Updated'].dt.day_name()
df.head()
```

```python
df['Update date_is_weekend'] = np.where(df['Update Day'].isin(['Sunday', 'Saturday']), "weekend","weekday")
df.head()
```

```python
df.sample(5)
```

```python
df['week of the year'] = df['Last Updated'].dt.isocalendar().week
df.head()
```

```python
df['quarter of year'] = df['Last Updated'].dt.quarter
df.head()
```

```python
df['semester'] = np.where(df['quarter of year'].isin([1,2]), 1, 2)
df.head()
```

# Elapsed time

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - df['Last Updated']
```

```python
(today - df['Last Updated']).dt.days
```

```python
# Months passed
# 1. Calculate the difference in days using np.timedelta64(1, 'D')
days_passed = (today - df['Last Updated']) / np.timedelta64(1, 'D')

# 2. Divide by 30.44 (average days per month) and round it
df['Months Passed'] = np.round(days_passed / 30.44, 0)
```
