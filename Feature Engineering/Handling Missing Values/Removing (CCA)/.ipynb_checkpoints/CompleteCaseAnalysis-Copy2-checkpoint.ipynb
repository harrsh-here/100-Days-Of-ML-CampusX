```python
missing_cols = [x for x in df.columns if df[x].isnull().mean()<0.05 and df[x].isnull().mean()>0]
missing_cols
```

```python
df['date_added'].isnull().mean()*100
```

```python
max(df[missing_cols].isnull().count())
```

```python
new_df = df[missing_cols].dropna()
```

```python
df.shape, new_df.shape
```

```python
# 1. Set the figure size FIRST
plt.figure(figsize=(12, 6))
for col in new_df.columns:
    # 2. Create the histogram (without figsize)
    plt.hist(new_df[col], density=True, edgecolor = 'black', color='#2b7bba')
    
    plt.tight_layout()
    # 3. Display the plot
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

for col in new_df.columns:
    # Skip if it actually is a numeric column
    if pd.api.types.is_numeric_dtype(df[col]):
        continue
        
    # 1. Calculate relative frequencies (proportions) for both datasets
    df_counts = df[col].value_counts(normalize=True).rename('Original')
    new_df_counts = new_df[col].value_counts(normalize=True).rename('After CCA')
    
    # 2. Combine them into a single temporary DataFrame for plotting
    df_compare = pd.concat([df_counts, new_df_counts], axis=1).fillna(0)
    
    # 3. Plot a grouped bar chart
    fig, ax = plt.subplots(figsize=(10, 5))
    df_compare.plot(kind='bar', ax=ax, color=['#d9534f', '#5cb85c'], width=0.8)
    
    # Styling
    ax.set_title(f'Categorical Distribution Comparison: {col}', fontsize=12, pad=15)
    ax.set_ylabel('Proportion (Percentage)')
    ax.set_xlabel(col)
    ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
    ax.legend(frameon=True)
    
    # Clean borders
    for spine in ['top', 'right']:
        ax.spines[spine].set_visible(False)
        
    plt.tight_layout()
    plt.show()
```

```python
col = missing_cols[0]
temp = pd.concat([
            # percentage of observations per category, original data
            df[col].value_counts() / len(df),

            # percentage of observations per category, cca data
            new_df[col].value_counts() / len(new_df)
        ],
        axis=1)
print("Column :", col)
# add column names
temp.columns = ['original', 'cca']

temp
```
